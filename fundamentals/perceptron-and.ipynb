```python
import numpy

b = 1
w = list()
def unitStep(v):
    if v > 0:
        return 1
    else:
        return 0
w=[1,2,1]
def myPerceptron(x1,x2,T):
    yhat = unitStep(b*w[0]+x1*w[1]+x2*w[2])
    err = T - yhat
    w[0] += err*b
    w[1] += err*x1
    w[2] += err*x2
for i in range(100):
    myPerceptron(0,0,0)
    myPerceptron(1,1,1)
    myPerceptron(1,0,1)
    myPerceptron(0,1,1)
for x1,x2 in [(0,0),(1,0),(0,1),(1,1)]:
    yhat = unitStep(b*w[0]+x1*w[1]+x2*w[2])
    print (x1,"OR",x2,"->",yhat)
```
