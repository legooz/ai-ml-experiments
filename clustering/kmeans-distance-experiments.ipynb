```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
import pandas as pd

# Load the iris dataset
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)


kmeans = KMeans(n_clusters=3)

# KMeans model 
kmeans.fit(df)


df['cluster'] = kmeans.labels_
```

```python
import matplotlib.pyplot as plt

# Visualization
x = df.iloc[:, [0, 1]].values


plt.scatter(x[df['cluster'] == 0, 0], x[df['cluster'] == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(x[df['cluster'] == 1, 0], x[df['cluster'] == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(x[df['cluster'] == 2, 0], x[df['cluster'] == 2, 1], s = 100, c = 'green', label = 'Cluster 3')


plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 200, c = 'yellow', label = 'Centroids')

plt.title('KMeans Clustering')
plt.xlabel('Sepal length')
plt.ylabel('Sepal width')
plt.legend()


plt.show()
from sklearn.metrics import silhouette_score

# Evaluation
silhouette_avg = silhouette_score(df, kmeans.labels_)
print("Silhouette score:", silhouette_avg)
```

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances
import pandas as pd
from sklearn.preprocessing import StandardScaler

# Load the iris dataset
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)

# Calculate Manhattan distance matrix
manhattan_dist = pairwise_distances(df_scaled, metric='manhattan')


kmeans = KMeans(n_clusters=3, random_state=0)

# Fit the KMeans model
kmeans.fit(manhattan_dist)

df['cluster'] = kmeans.labels_
```

```python
plt.scatter(df_scaled[df['cluster'] == 0, 0], df_scaled[df['cluster'] == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(df_scaled[df['cluster'] == 1, 0], df_scaled[df['cluster'] == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(df_scaled[df['cluster'] == 2, 0], df_scaled[df['cluster'] == 2, 1], s = 100, c = 'green', label = 'Cluster 3')

plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 200, c = 'yellow', label = 'Centroids')

plt.title('KMeans Clustering')
plt.xlabel('Scaled Sepal length')
plt.ylabel('Scaled Sepal width')
plt.legend()
plt.show()
from sklearn.metrics import silhouette_score
# Evaluate 
silhouette_avg = silhouette_score(df, kmeans.labels_)
print("Silhouette score:", silhouette_avg)
```

```python
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import pairwise_distances
import pandas as pd
from sklearn.preprocessing import StandardScaler
# Load the iris dataset
iris = load_iris()

df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
scaler = StandardScaler()
df_scaled = scaler.fit_transform(df)
# Calculate cosine similarity matrix
cosine_sim = 1 - pairwise_distances(df, metric='cosine')


kmeans = KMeans(n_clusters=3, random_state=0)

# Fit the KMeans model 
kmeans.fit(cosine_sim)


df['cluster'] = kmeans.labels_
```

```python
import matplotlib.pyplot as plt
plt.scatter(df_scaled[df['cluster'] == 0, 0], df_scaled[df['cluster'] == 0, 1], s = 100, c = 'red', label = 'Cluster 1')
plt.scatter(df_scaled[df['cluster'] == 1, 0], df_scaled[df['cluster'] == 1, 1], s = 100, c = 'blue', label = 'Cluster 2')
plt.scatter(df_scaled[df['cluster'] == 2, 0], df_scaled[df['cluster'] == 2, 1], s = 100, c = 'green', label = 'Cluster 3')

plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s = 200, c = 'yellow', label = 'Centroids')

plt.title('KMeans Clustering')
plt.xlabel('Scaled Sepal length')
plt.ylabel('Scaled Sepal width')
plt.legend()
plt.show()

from sklearn.metrics import silhouette_score
silhouette_avg = silhouette_score(df, kmeans.labels_)
print("Silhouette score:", silhouette_avg)
```
